# UCI-SECOM Dataset EDA

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from uci_secom_outlier_detection.data_loader import load_dataset
```

```python
df, y = load_dataset()
stats_df = df.describe()
```

```python
print(df.head())
print(df.shape)
print(stats_df)
```

```python
print(y.head(3))
y.shape
```

```python
missing_pct = df.isna().mean().sort_values(ascending=False)
print(missing_pct.head(10))
print((missing_pct > 0.20).sum()) # how many have >20% missing values
variances = df.var()
print((variances < 1e-6).sum()) # how many columns are almost constant
```

There seems to be some subgroups of features that have the same exact percentage. This is unlikely to be due to chance, and might be indicating a shared a common mode of failure.

```python
fail_rate = (y == 1).mean()
print(f"{fail_rate:.2%} ({(y == 1).sum()} fails out of {len(y)})")
```

104 failts out of 1567 matches the dataset description.
